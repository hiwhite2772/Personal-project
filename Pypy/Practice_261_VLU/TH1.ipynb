```python
def phanloai_huyetap(tamthu, tamtruong):
    if tamthu >= 180 or tamtruong >= 110:
        return "THA độ 3 (nặng)"
    elif (160 <= tamthu < 180) or (100 <= tamtruong < 110):
        return "THA độ 2 (trung bình)"
    elif (140 <= tamthu < 160) or (90 <= tamtruong < 100):
        return "THA độ 1 (nhẹ)"
    elif (130 <= tamthu < 140) or (85 <= tamtruong < 90):
        return "Bình thường cao"
    elif (120 <= tamthu < 130) or (80 <= tamtruong < 85):
        return "Bình thường (normal)"
    elif (tamthu < 120) and (tamtruong < 80):
        return "Tốt nhất (optimal)"
    elif (tamthu < 90) and (tamtruong < 60):
        return "Thấp (hypotension)"
    elif (tamthu >= 180) and (tamtruong < 90):
        return "THA tâm thu đơn độc"
    else:
        return "Không xác định"
phanloai_huyetap(55, 66)
```
